# Creating Virtual Envionment


For Every Project it is necessary to create its own venv(virtual environment) where you can install all the packages and run you project
you can create you venv using the following command

`conda create -p venv python`

After Creating the venv we need to activate it using this following command

`conda activate venv`
        or
`conda activate <venv path>`

you can deactivate venv using

`conda deactivate <venv path>`

for vscode we need to install a kernel selector to select our kernal

`pip install ipykernel`

now create .ipynb file and select the venv(Python 3.10.19) you created and run the project



# Different ways to create virtual environment 

## 1. Using simple Python
If someone does not have anaconda installed. then he is not able to run conda commands
In this phase he must use these commands by using this commands we can create venv just by using the python

`python -m venv <env name>`

After Executing this you can see your env folder in that you have folder called Scripts which contains a file activate. which is used to activate this enviromnet

`<env name> \Scripts\activate`

To deactivate the venv just run
`deactivate`


## 2. Using pip package
We can install virtualenv package to create virual enviroment
`pip install virtualenv`

To create virtual enviroment we use
`virtualenv -p python3 <env name>`

To Activate virtual enviroment use the command
`<env name>\Scripts\activate`

To deactivate the venv just run
`deactivate`

# PIP


Create a requirements.txt file and add all the packages you will be using.

you can directly install the packages using command

`pip install -r requirements.txt`


You Can Install Packages Using Pip (python installs packages)

| Command                           | Description                      |
| --------------------------------- | -------------------------------- |
| `pip install package_name`        | Install a package                |
| `pip install -U package_name`     | Upgrade a package                |
| `pip uninstall package_name`      | Remove a package                 |
| `pip list`                        | Show installed packages          |
| `pip show package_name`           | Show details of a package        |
| `pip freeze > requirements.txt`   | Save all installed packages      |
| `pip install -r requirements.txt` | Install all packages from a file |
