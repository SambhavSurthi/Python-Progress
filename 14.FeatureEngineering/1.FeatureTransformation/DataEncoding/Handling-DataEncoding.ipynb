```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
```

```python
df=pd.DataFrame(
    {
        'color':['red','blue','green','blue','green','red','blue','green','red']
    }
)
```

```python
df.head()
```

```python
ohe=OneHotEncoder()
```

```python
encoded=ohe.fit_transform(df[['color']]).toarray()
df=pd.DataFrame(data=encoded,columns=ohe.get_feature_names_out())
```

```python
df
```

# Encode all the categorical data in seaborn tips dataset

```python
import seaborn as sns

dataset=sns.load_dataset('tips')
```

```python
dataset
```

```python
dataset.isna().sum()
```

```python
dataset.head()
```

```python
from sklearn.preprocessing import OneHotEncoder
import pandas as pd
encoder=OneHotEncoder()
sex_encoded=encoder.fit_transform(dataset[['sex']]).toarray()
sex_df=pd.DataFrame(sex_encoded,columns=encoder.get_feature_names_out())
dataset=dataset.join(sex_df,how='inner')

```

```python
dataset
```

```python
smoker_encoded=encoder.fit_transform(dataset[['smoker']]).toarray()
smoker_df=pd.DataFrame(smoker_encoded,columns=encoder.get_feature_names_out())
dataset=dataset.join(smoker_df,how='inner')
```

```python
dataset
```

```python
day_encoded=encoder.fit_transform(dataset[['day']]).toarray()
day_df=pd.DataFrame(day_encoded,columns=encoder.get_feature_names_out())
dataset=dataset.join(day_df,how='inner')
```

```python
dataset
```

```python
time_encoded=encoder.fit_transform(dataset[['time']]).toarray()
time_df=pd.DataFrame(time_encoded,columns=encoder.get_feature_names_out())
dataset=dataset.join(time_df,how='inner')
```

```python
dataset
```

# Label Encoding

```python
df=pd.DataFrame(
    {
        'color':['red','blue','green','blue','green','red','blue','green','red']
    }
)
df.head()
```

```python
from sklearn.preprocessing import LabelEncoder

lbl_encoder=LabelEncoder()
lbl_encoded=lbl_encoder.fit_transform(df['color'])
```
