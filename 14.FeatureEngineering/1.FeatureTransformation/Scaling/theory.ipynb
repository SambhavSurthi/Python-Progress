# Feature Scaling

## What is Feature Scaling?

Feature scaling is the process of bringing numerical features to a similar scale so that no single feature dominates learning just because of its magnitude.

**In simple words:** Feature scaling makes features comparable.

## Why Do We Need Feature Scaling?

Most ML algorithms either:
- Use distance calculations, or
- Use gradient descent

**If features are on different scales, the model:**
- Learns slowly
- Gives biased importance to large-scale features
- May converge to a wrong or suboptimal solution

### Example

| Feature | Range |
|---------|-------|
| Age | 18–60 |
| Salary | ₹20,000–₹2,00,000 |

**Problem:** Salary dominates distance & gradients, while Age becomes almost irrelevant.

---

## Types of Feature Scaling

### 1. Standardization (Z-score Scaling)

$$x' = \frac{x - \mu}{\sigma}$$

| Aspect | Details |
|--------|---------|
| **What it does** | Mean → 0, Std → 1, No fixed range |
| **When to use** | Data is roughly normal, Algorithms using gradients or distance |
| **Used by** | Linear Regression, Logistic Regression, SVM, PCA, KNN |

### 2. Normalization (Min–Max Scaling)

$$x' = \frac{x - x_{min}}{x_{max} - x_{min}}$$

| Aspect | Details |
|--------|---------|
| **What it does** | Scales data to [0, 1], Preserves shape |
| **When to use** | Neural Networks, Image data, Bounded inputs required |

### 3. Robust Scaling

$$x' = \frac{x - \text{median}}{IQR}$$

| Aspect | Details |
|--------|---------|
| **Why** | Resistant to outliers |
| **Use when** | Many extreme values |

### 4. MaxAbs Scaling

$$x' = \frac{x}{|x_{max}|}$$

| Aspect | Details |
|--------|---------|
| **Why** | Keeps zero intact |
| **Use when** | Sparse data |

### 5. Unit Vector Scaling (L2 Normalization)

$$x' = \frac{x}{||x||}$$

| Aspect | Details |
|--------|---------|
| **Why** | Direction matters more than magnitude |
| **Used in** | Text (TF-IDF), Cosine similarity |

---

## What Feature Scaling Does NOT Fix

Feature scaling cannot address:
- ❌ Skewed distributions
- ❌ Non-linearity
- ❌ Outliers (except robust scaler)

**For these issues, use:**
- Log/Power transformation
- Feature engineering

## Decision Guide: When to Use What?

| Situation | Recommended Scaler |
|-----------|-------------------|
| Different units | Standardization |
| Neural network | Normalization |
| Outliers present | RobustScaler |
| Sparse data | MaxAbsScaler |
| Text data | L2 Normalization |

# Normalization vs Standardization

Both are feature scaling techniques, but they solve different problems and are used in different situations.

---

## 1. Normalization (Min–Max Scaling)

### What It Does

Scales data into a fixed range, usually [0, 1].

$$x' = \frac{x - x_{min}}{x_{max} - x_{min}}$$

### Properties

- Range is fixed to [0, 1]
- Shape of distribution is preserved
- Sensitive to outliers

### When to Use

- Neural Networks
- Image/pixel data
- When algorithm expects bounded inputs

---

## 2. Standardization (Z-score Scaling)

### What It Does

Centers data around 0 with unit variance.

$$x' = \frac{x - \mu}{\sigma}$$

### Properties

- Mean = 0
- Standard Deviation = 1
- No fixed range
- Less sensitive to outliers (but still affected)

### When to Use

- Linear/Logistic Regression
- SVM
- PCA
- KNN
- Gradient-based models

---

## Side-by-Side Comparison

| Aspect | Normalization | Standardization |
|--------|---------------|-----------------|
| **Range** | Fixed (0–1) | Not fixed |
| **Mean** | Not centered | Centered at 0 |
| **Standard Deviation** | Not fixed | 1 |
| **Outlier Sensitivity** | High | Moderate |
| **Distribution Shape** | Preserved | Preserved |
| **Best Use Case** | Neural Networks | Most ML algorithms |

---

## Quick Decision Guide

1. **Neural network?** → Normalization
2. **Distance/gradient-based model?** → Standardization
3. **Outliers present?** → RobustScaler (neither)
4. **Tree models?** → No scaling needed

---

## Common Mistakes

| ❌ Wrong | ✅ Correct |
|---------|-----------|
| Using normalization thinking it "makes data normal" | Normalization only rescales values, does not change distribution |

---

## Interview One-Liner

> **Normalization** rescales data to a fixed range, typically [0,1], while **Standardization** rescales data to have zero mean and unit variance. Normalization is preferred for neural networks, whereas standardization is preferred for most classical machine learning algorithms.
