```python
# Import All the necessary Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
# Create Dataset
np.random.seed(23)

mu_vec1 = np.array([0, 0, 0])
cov_mat1 = np.array([[1,0,0],[0,1,0],[0,0,1]])
class1_sample = np.random.multivariate_normal(mu_vec1, cov_mat1, 20)

df = pd.DataFrame(class1_sample, columns=['feature1','feature2','feature3'])
df['target'] = 1

mu_vec2 = np.array([1, 1, 1])
cov_mat2 = np.array([[1,0,0],[0,1,0],[0,0,1]])
class2_sample = np.random.multivariate_normal(mu_vec2, cov_mat2, 20)

df1 = pd.DataFrame(class2_sample, columns=['feature1','feature2','feature3'])
df1['target'] = 0

# ✅ Correct way to combine DataFrames
df = pd.concat([df, df1], ignore_index=True)

# Shuffle dataset
df = df.sample(frac=1, random_state=42).reset_index(drop=True)
```

# PCA Without any Libraries
1. **Standardize data** (important!)
2. Compute **covariance matrix** (\Sigma)
3. Compute **eigenvectors & eigenvalues** of (\Sigma)

   * Eigenvectors → directions (PCs)
   * Eigenvalues → variance captured
4. Sort by eigenvalues (descending)
5. Project data onto top (k) eigenvectors

```python
X=df.iloc[:,0:3]
y=df.iloc[:,-1]
```

```python
X
```

```python
y
```

```python
# Step 1: Standardize the Values
from sklearn.preprocessing import StandardScaler

scaler=StandardScaler()
X=scaler.fit_transform(X=X)
```

```python
#Convert np array to Dataframe
X_Scaled=pd.DataFrame(X,columns=['feature1','feature2','feature3'])
```

```python
# Construct Covarinece Matrix
covarienceMatrix=X_Scaled.cov()
```

```python
# Extract eigen values and eigen vectors

eigenValues,eigenVectors=np.linalg.eig(covarienceMatrix)
```

```python
eigenValues
```

```python
eigenVectors
```

```python
# As We are converting 3D to 2D (Converting 3 features to 2 features), we select top 2 highest eigen values and eigen vectors these are principal components
pc=eigenVectors[:2]
```

```python
pc
```

```python
# For the final step. Calculate the dot product of the row and transposed principal component

transformedDF=np.dot(df.iloc[:,:3],pc.T)
```

```python
# New Trasformed Dataset
transformedDF
```

```python
# Now Convert this to dataset
new_df = pd.DataFrame(transformedDF,columns=['PC1','PC2'])
new_df['target'] = df['target'].values
new_df.head()
```

```python
# New 2D Diagram
new_df['target'] = new_df['target'].astype('str')
fig = px.scatter(x=new_df['PC1'],
                 y=new_df['PC2'],
                 color=new_df['target'],
                 color_discrete_sequence=px.colors.qualitative.G10
                )
fig.show()
```
