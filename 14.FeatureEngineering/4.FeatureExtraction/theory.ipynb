# 🧠 1) Variance

## 📌 What is variance?

> **Variance measures how spread out data is around its mean.**

If values are far from the mean → high variance
If values are close to the mean → low variance

---

## 🧮 Formula

For a feature (X = {x_1,\dots,x_n}) with mean (\mu):

$$\mathrm{Var}(X) = \frac{1}{n}\sum_{i=1}^{n}(x_i - \mu)^2$$

(sample variance often uses (\frac{1}{n-1}))

---

## 🧠 Intuition

* Squaring avoids positive/negative cancellation
* Emphasizes large deviations
* Units are **squared** (important!)

---

## 🎯 Why it matters

* Tells how informative a feature is
* High variance often means more signal (not always!)
* PCA uses variance to decide what to keep

---

# 🧠 2) Covariance

## 📌 What is covariance?

> **Covariance measures how two variables change together.**

---

## 🧮 Formula

For features (X) and (Y):

$$\mathrm{Cov}(X,Y) = \frac{1}{n}\sum_{i=1}^{n}(x_i-\mu_X)(y_i-\mu_Y)$$

---

## 🧠 Interpretation

* **Positive** → increase together
* **Negative** → one up, other down
* **Zero** → no linear relationship

---

## ⚠️ Limitation

* Scale-dependent (hard to compare)
* Correlation fixes this by normalizing

---

## 📦 Covariance Matrix

For many features:

$$\Sigma = \begin{bmatrix} \mathrm{Var}(X_1) & \mathrm{Cov}(X_1,X_2) & \dots \\ \mathrm{Cov}(X_2,X_1) & \mathrm{Var}(X_2) & \dots \\ \vdots & \vdots & \ddots \end{bmatrix}$$

This matrix is the **starting point of PCA**.

---

# 🧠 3) Eigenvalues & Eigenvectors

## 📌 Big idea (plain English)

> **Eigenvectors are special directions.
> Eigenvalues tell how important those directions are.**

---

## 🧮 Mathematical definition

For a matrix (A):

$$A\mathbf{v} = \lambda \mathbf{v}$$

* (\mathbf{v}) = eigenvector (direction)
* (\lambda) = eigenvalue (scale/strength)

---

## 🧠 Geometrical intuition

Applying matrix (A):

* Most vectors change direction
* **Eigenvectors keep their direction**
* Eigenvalue tells how much stretching happens along that direction

---

## 🎯 In statistics / PCA

When (A) is the **covariance matrix**:

* **Eigenvectors** → principal directions (new axes)
* **Eigenvalues** → variance along those directions

---

# 🧠 4) Explained Variance

## 📌 What is explained variance?

> **How much of the total data variance is captured by a component (direction).**

---

## 🧮 Formula

If eigenvalues are (\lambda_1,\lambda_2,\dots):

$$\text{Explained Variance Ratio of PC}_k = \frac{\lambda_k}{\sum_i \lambda_i}$$

---

## 🧠 Intuition

* Big eigenvalue → component captures lots of spread
* Small eigenvalue → mostly noise

---

## 🎯 Why it matters

* Decide **how many components to keep** in PCA
* Common rule: keep enough PCs to explain **90–95%** variance

---

# 🧠 5) Eigenvectors (in PCA context)

## 📌 What are eigenvectors here?

> **They define the new coordinate axes (principal components).**

Properties:

* Orthogonal (perpendicular)
* Uncorrelated
* Ordered by importance (via eigenvalues)

---

## 🧭 Geometrical view

* PC1 = direction of **maximum variance**
* PC2 = next max variance, perpendicular to PC1
* And so on…

---

# 🧠 6) Unit Vector

## 📌 What is a unit vector?

> **A vector with length (magnitude) = 1.**

---

## 🧮 Formula

For vector (\mathbf{v}):

$$\hat{\mathbf{v}} = \frac{\mathbf{v}}{|\mathbf{v}|}$$

---

## 🧠 Why we use unit vectors

* Represent **direction only**
* Remove scale effects
* Eigenvectors are usually **normalized to unit length**

---

## 🎯 In PCA

* Eigenvectors are converted to **unit vectors**
* Projection becomes clean and comparable

---

# 🧠 How ALL of these connect (VERY IMPORTANT)

### Step-by-step PCA pipeline

1. **Center data** (mean = 0)
2. Compute **covariance matrix**
3. Find **eigenvectors & eigenvalues**
4. Eigenvectors → directions (PCs)
5. Eigenvalues → variance along PCs
6. **Explained variance** decides how many PCs to keep
7. Eigenvectors are **unit vectors**
8. Project data onto them

---

# 🧠 One-Table Summary

| Term               | Meaning                                |
| ------------------ | -------------------------------------- |
| Variance           | Spread of a single variable            |
| Covariance         | Joint variation of two variables       |
| Covariance Matrix  | Variance + covariance of all features  |
| Eigenvector        | Special direction that doesn't rotate  |
| Eigenvalue         | Strength/variance along that direction |
| Explained Variance | Fraction of total variance captured    |
| Unit Vector        | Direction with magnitude 1             |

---

# 🎯 Interview One-Liners (GOLD)

* **Variance** measures spread of data around the mean.
* **Covariance** measures how two variables move together.
* **Eigenvectors** define principal directions of data.
* **Eigenvalues** quantify variance along those directions.
* **Explained variance** tells how much information a component retains.
* **Unit vectors** represent direction without magnitude.

---

# 🧠 Final Mental Model

```
Data spread → Variance
Joint spread → Covariance
All spreads → Covariance Matrix
Special directions → Eigenvectors
Importance → Eigenvalues
Keep best → Explained Variance
Normalize → Unit Vectors
```

## 📌 What is the Curse of Dimensionality?

> **As the number of features (dimensions) increases, data becomes sparse, distances lose meaning, models need exponentially more data, and performance degrades.**

---

## ❓ Why does it happen?

Because **volume grows exponentially** with dimensions, but **data grows linearly**.

* 1D interval ([0,1]): length = 1
* 2D square ([0,1]^2): area = 1
* 10D hypercube ([0,1]^{10}): volume = 1 (still), but **points are extremely sparse** inside it

To “fill” the space even modestly, you’d need **exponentially many samples**.

---

## 🔍 What problems does it cause?

### 1️⃣ Distance concentration

In high dimensions, **nearest and farthest points are almost equally far**.

**Effect:**

* KNN, K-means, SVM (distance-based) struggle.
* “Nearest neighbor” stops being meaningful.

---

### 2️⃣ Sparsity

Most regions have **no data**.

**Effect:**

* Models overfit (memorize noise).
* Local patterns are unreliable.

---

### 3️⃣ Overfitting & variance explosion

More features ⇒ more parameters ⇒ higher variance.

**Effect:**

* Great training score, poor test score.

---

### 4️⃣ Computational blow-up

* Time and memory increase rapidly.
* Optimization becomes unstable.

---

## 🧮 Mathematical intuition (simple)

Let (d) be dimensions, (n) points uniformly in ([0,1]^d).

* Average distance between random points **increases** with (d).
* Ratio (\frac{\text{max distance} - \text{min distance}}{\text{min distance}} \to 0) as (d \to \infty).

**Meaning:** distances become indistinguishable.

---

## 🧭 Geometrical intuition

* In low dimensions, points cluster and neighborhoods exist.
* In high dimensions, points lie near the **edges/corners** of space.
* The “interior” becomes negligible.

---

## 🧯 How to solve / mitigate it

| Strategy                     | How it helps                                  |
| ---------------------------- | --------------------------------------------- |
| **Feature Selection**        | Remove irrelevant/redundant features          |
| **Feature Extraction (PCA)** | Compress dimensions while keeping information |
| **Regularization (L1/L2)**   | Penalize complexity                           |
| **More data**                | Counter sparsity (often expensive)            |
| **Use tree models**          | Less distance-sensitive                       |
| **Domain features**          | Fewer, stronger features                      |

---

## 🎯 One-line takeaway

> **High dimensions dilute data; reduce dimensions or models lose meaning.**

## 📌 What is PCA?

> **PCA is an unsupervised linear technique that reduces dimensionality by projecting data onto orthogonal directions of maximum variance.**

---

## ❓ What problems does PCA solve?

* Curse of dimensionality
* Multicollinearity (correlated features)
* Noise reduction
* Visualization (high-D → 2D/3D)
* Faster, more stable training

---

## 🧠 What does PCA do (conceptually)?

* Finds **new axes** (principal components)
* Each axis captures **as much variance as possible**
* Axes are **orthogonal (uncorrelated)**
* Keeps top (k) axes, drops the rest

---

## 🧭 Geometrical intuition

Imagine points forming an **elongated cloud**.

1. **PC1**: direction of **maximum spread** (longest axis)
2. **PC2**: next largest spread, **perpendicular** to PC1
3. Project points onto first (k) PCs

**Result:** fewer dimensions, minimal information loss.

---

## 🧮 Mathematical intuition (clean & practical)

1. **Standardize data** (important!)
2. Compute **covariance matrix** (\Sigma)
3. Compute **eigenvectors & eigenvalues** of (\Sigma)

   * Eigenvectors → directions (PCs)
   * Eigenvalues → variance captured
4. Sort by eigenvalues (descending)
5. Project data onto top (k) eigenvectors

**Explained Variance Ratio** tells how much information each PC keeps.

---

## 🧪 Why standardization before PCA?

PCA is **variance-driven**.
Without scaling, features with larger units dominate PCs.

---

## 🧩 What PCA does NOT do

* ❌ Does not use labels (unsupervised)
* ❌ Does not keep original feature meaning
* ❌ Does not capture non-linear structure (use Kernel PCA/Autoencoders)

---

## ⏱️ When should you use PCA?

✔ Many correlated numerical features
✔ Distance-based or linear models
✔ Visualization needed
✔ Noise reduction helps
✔ Faster training required

❌ If interpretability of original features is crucial
❌ For tree-based models (often unnecessary)

---

## 📊 How PCA helps models

| Model Type        | Benefit                   |
| ----------------- | ------------------------- |
| Linear / Logistic | Removes multicollinearity |
| KNN / K-means     | Meaningful distances      |
| SVM               | Stable margins            |
| NN                | Faster convergence        |

---

## ⚠️ Common mistakes

* Skipping standardization
* Applying PCA before train/test split (leakage)
* Keeping too many components
* Expecting interpretability

---

## 🧠 PCA vs Feature Selection

| Aspect                  | PCA | Feature Selection |
| ----------------------- | --- | ----------------- |
| Keeps original features | ❌   | ✅                 |
| Reduces dimensionality  | ✅   | ✅                 |
| Interpretability        | ❌   | ✅                 |
| Handles correlation     | ✅   | ❌ (often)         |

---

## 🎯 Choosing number of components

* **Explained variance** (e.g., keep 95%)
* **Elbow plot**
* **Cross-validation**

---

## 🏁 FINAL MENTAL MODELS

### Curse of Dimensionality

```
More dimensions
   ↓
Sparser data
   ↓
Distances lose meaning
   ↓
Overfitting & poor generalization
```

### PCA

```
Correlated features
   ↓
Orthogonal directions (PCs)
   ↓
Keep max variance
   ↓
Fewer dimensions, less noise
```

---

## 🔑 FINAL ONE-LINERS (INTERVIEW GOLD)

* **Curse of dimensionality**: performance degrades as dimensions increase due to sparsity and distance concentration.
* **PCA**: projects data onto orthogonal directions of maximum variance to reduce dimensions with minimal information loss.
