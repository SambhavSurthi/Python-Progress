# Handling Outliers: 5 Number Summary, IQR, lowerfence, higherfence, Boxplot

```python
import numpy as np

lst_marks=[15,14,2,5,1,4,0,44,88,2,77,11,55,11,77,11,101,99,10,45,15,11,48,85,41,46,49]
minimum_val,Q1,median_val,Q3,maximum_val=np.quantile(lst_marks,[0.0,0.25,0.50,0.75,1.0])
```

```python
print(minimum_val,Q1,median_val,Q3,maximum_val)
```

```python
IQR=Q3-Q1
print(IQR)
```

```python
lowerfence=Q1-1.5*IQR
higherfence=Q3+1.5*IQR

print(lowerfence,higherfence)
```

## Box Plot

```python
import seaborn as sns

sns.boxplot(lst_marks)
```
