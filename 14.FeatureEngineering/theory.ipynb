# 🧠 FEATURE ENGINEERING — COMPLETE MASTER GUIDE (A–Z)

## 📌 What is Feature Engineering?

> **Feature Engineering is the process of transforming raw data into meaningful features that improve a machine learning model’s performance.**

In simple words:

> **Better features → better models (often more than better algorithms).**

---

## 🧩 Why Feature Engineering is CRITICAL

Algorithms **do not understand meaning**.
They only understand **numbers & patterns**.

Raw data usually has:

* Noise
* Missing values
* Wrong scales
* Irrelevant features
* Hidden relationships

👉 Feature engineering **bridges the gap between raw data and model intelligence**.

---

# 🧠 BIG PICTURE: FEATURE ENGINEERING CATEGORIES

Feature Engineering has **4 major pillars**:

```
Feature Engineering
│
├── 1. Feature Transformation
│
├── 2. Feature Construction
│
├── 3. Feature Selection
│
└── 4. Feature Extraction
```

We’ll cover **everything inside each one**.

---

# 1️⃣ FEATURE TRANSFORMATION

## 📌 What is Feature Transformation?

> **Changing the representation or scale of existing features without creating new ones.**

You are not creating new information —
you are **making existing information usable**.

---

## 🔹 1.1 Missing Value Imputation

### 📌 What it is

Handling missing (`NaN`) values so models can work.

### ❓ Why needed

Most ML models **cannot handle NaNs**.

---

### 🔸 Types of Imputation

#### (A) Simple Imputation

* Mean
* Median (best for skewed data)
* Mode (categorical)

#### (B) Constant Value

* `0`, `Unknown`, `Missing`

#### (C) Statistical / Advanced

* KNN Imputer
* Iterative Imputer
* **MICE** (explained later)

#### (D) Missing Indicator

* Add a new column: `is_missing`

✔ Often improves performance

---

## 🔹 1.2 Handling Categorical Data

### 📌 What it is

Converting categories into numeric form.

### ❓ Why needed

Models cannot understand strings.

---

### 🔸 Encoding Techniques

| Encoding           | When to Use           |
| ------------------ | --------------------- |
| Label Encoding     | Ordinal categories    |
| One-Hot Encoding   | Nominal categories    |
| Ordinal Encoding   | Ordered categories    |
| Target Encoding    | High-cardinality      |
| Frequency Encoding | Rare categories       |
| Binary Encoding    | Very high cardinality |

---

## 🔹 1.3 Feature Scaling

### 📌 What it is

Bringing features to a similar scale.

### ❓ Why needed

Distance & gradient-based models are scale-sensitive.

---

### 🔸 Types of Scaling

| Technique                 | Use Case           |
| ------------------------- | ------------------ |
| Standardization (Z-score) | Linear models, SVM |
| Normalization (MinMax)    | Neural Networks    |
| Robust Scaling            | Outliers present   |
| MaxAbs Scaling            | Sparse data        |

---

## 🔹 1.4 Feature Transformation (Distribution-based)

### 📌 What it is

Changing **shape of data distribution**.

---

### 🔸 Types

| Technique       | Purpose                |
| --------------- | ---------------------- |
| Log Transform   | Reduce right skew      |
| Power Transform | Normalize distribution |
| Box-Cox         | Positive data          |
| Yeo-Johnson     | Any real values        |
| Reciprocal      | Compress large values  |

---

## 🔹 1.5 Encoding Numerical Features

This is **often overlooked but very powerful**.

### 🔸 Discretization / Binning

* Equal Width
* Equal Frequency (Quantile)
* K-Means Binning
* Decision Tree Binning (Supervised)
* Custom Binning

### 🔸 Binarization

* Convert numeric → 0/1 using threshold

---

# 2️⃣ FEATURE CONSTRUCTION

## 📌 What is Feature Construction?

> **Creating new features from existing features.**

This is where **domain knowledge shines**.

---

## 🔹 2.1 Feature Splitting

### 📌 What it is

Breaking one feature into multiple parts.

### Examples:

* Date → Year, Month, Day
* Name → First name, Last name
* Address → City, State, Country

✔ Helps models capture granular patterns

---

## 🔹 2.2 Feature Combination

### 📌 What it is

Combining multiple features into one.

### Examples:

* `TotalSpend = Price × Quantity`
* `BMI = Weight / Height²`

---

## 🔹 2.3 Interaction Features

### 📌 What it is

Capturing interaction between features.

Example:

```
Income × Education
```

Useful for:

* Linear models
* Polynomial regression

---

## 🔹 2.4 Aggregation Features

Used heavily in:

* Time-series
* Transactional data

Examples:

* Avg purchase per user
* Max login per day
* Rolling mean

---

# 3️⃣ FEATURE SELECTION

## 📌 What is Feature Selection?

> **Choosing the most useful features and removing the rest.**

---

## ❓ Why needed

* Reduce overfitting
* Faster training
* Better generalization
* Improved interpretability

---

## 🔹 Types of Feature Selection

---

### 🔸 3.1 Filter Methods (Model Independent)

* Correlation
* Variance Threshold
* Chi-Square
* ANOVA

✔ Fast
❌ Ignores model

---

### 🔸 3.2 Wrapper Methods (Model Based)

* Forward Selection
* Backward Elimination
* Recursive Feature Elimination (RFE)

✔ Accurate
❌ Computationally expensive

---

### 🔸 3.3 Embedded Methods (Best of both)

* Lasso (L1)
* Ridge (L2)
* ElasticNet
* Tree Feature Importance

✔ Most used in practice

---

# 4️⃣ FEATURE EXTRACTION

## 📌 What is Feature Extraction?

> **Reducing dimensionality by transforming data into a lower-dimensional space.**

Unlike feature selection:

* Original features are **not preserved**
* New abstract features are created

---

## 🔹 4.1 PCA (Principal Component Analysis)

### 📌 What it does

* Converts correlated features into uncorrelated components
* Maximizes variance

### When to use

* High-dimensional numeric data
* Multicollinearity
* Visualization

❌ Not interpretable

---

## 🔹 4.2 MICE (Multiple Imputation by Chained Equations)

### 📌 What it does

* Advanced **missing value imputation**
* Uses other features to predict missing values

### Why it’s in Feature Extraction

* Creates new inferred values
* Uses multivariate relationships

---

## 🔹 Other Feature Extraction Techniques

| Technique          | Use Case                            |
| ------------------ | ----------------------------------- |
| LDA                | Supervised dimensionality reduction |
| Autoencoders       | Deep learning                       |
| t-SNE / UMAP       | Visualization                       |
| Word Embeddings    | NLP                                 |
| Image CNN Features | Computer Vision                     |

---

# 🧠 Feature Engineering vs Feature Extraction vs Selection

| Aspect                   | Engineering | Selection | Extraction |
| ------------------------ | ----------- | --------- | ---------- |
| Creates new features     | ✅           | ❌         | ✅          |
| Removes features         | ❌           | ✅         | ❌          |
| Interpretability         | High        | High      | Low        |
| Dimensionality reduction | Sometimes   | Yes       | Yes        |

---

# ⚠️ IMPORTANT RULES (EXAM + REAL WORLD)

1. Fit transformations **only on training data**
2. Apply same transformation to test data
3. Use Pipelines to avoid leakage
4. Trees need less feature engineering
5. Linear models need more

---

# 🎯 FINAL ONE-LINE SUMMARY (GOLD)



# ColumnTransformer — Complete Guide

## Definition

**ColumnTransformer** is a sklearn utility that allows you to:

- Apply different preprocessing steps to different columns simultaneously
- Handle multiple feature types in one unified workflow

---

## Why do we need ColumnTransformer?

### Without ColumnTransformer ❌

You would have to:

- Split columns manually
- Transform separately
- Recombine arrays
- Risk column mismatch & data leakage
- Messy and error-prone

### With ColumnTransformer ✅

- One clean object
- Column-wise preprocessing
- Safe for train/test split
- Pipeline-friendly
- Production-ready

---

## Real-World Example

### Dataset with Mixed Feature Types

| Feature | Type | Preprocessing |
|---|---|---|
| Age | Numerical | Binning |
| Fare | Numerical | Log + Scaling |
| Sex | Categorical | OneHot |
| City | Categorical | Target Encoding |

**ColumnTransformer handles ALL this in one step.**

---

## Implementation Example

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder, KBinsDiscretizer

preprocessor = ColumnTransformer([
    ('age_bin', KBinsDiscretizer(n_bins=5), ['Age']),
    ('fare_scale', StandardScaler(), ['Fare']),
    ('sex_ohe', OneHotEncoder(), ['Sex'])
], remainder='passthrough')
```

### Anatomy of ColumnTransformer

```python
ColumnTransformer([
    (name, transformer, columns),
    (name, transformer, columns),
    ...
], remainder='drop')
```

- **name**: String identifier for this transformation step
- **transformer**: sklearn transformer object (fit/transform methods)
- **columns**: List of column names or indices to apply this transformer to
- **remainder**: What to do with untransformed columns

---

## How ColumnTransformer Helps ML

| Problem | How it helps |
|---|---|
| Different feature types | Handles each correctly |
| Avoids data leakage | Fits only on training data |
| Pipeline integration | Cross-validation safe |
| Cleaner code | More maintainable |
| Reproducibility | Production-ready |

---

## Important Defaults ⚠️

### remainder Parameter

The `remainder` parameter is **very important**:

```python
remainder='drop'        # Removes all untouched columns (DEFAULT)
remainder='passthrough' # Keeps untouched columns as-is
```

### Critical Point

**If you don't specify `remainder` parameter:**

```python
remainder='drop'  # This is the default
```

👉 **All other columns are removed!**

### When to Use Each

| Value | Use Case | Example |
|---|---|---|
| `'drop'` | Only want transformed features | Feature selection focused |
| `'passthrough'` | Want to keep additional features | Include IDs, untouched numerics |

---

## Integration with Pipeline

### Safe Preprocessing + Model Training

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

pipe = Pipeline([
    ('preprocess', preprocessor),
    ('model', LogisticRegression())
])
```

### How This Prevents Data Leakage

1. **During training**: Fit transformers on training data only
2. **During CV**: Each fold fits separately
3. **During test**: Apply same transformers (no refitting)
4. **Result**: No information leaks from test to training

---

## Complete Pipeline Example

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.ensemble import RandomForestClassifier

# Define numerical and categorical columns
numeric_cols = ['Age', 'Fare']
categorical_cols = ['Sex', 'City']

# Create transformers
numeric_transformer = StandardScaler()
categorical_transformer = OneHotEncoder(drop='first', sparse_output=False)

# Create ColumnTransformer
preprocessor = ColumnTransformer([
    ('num', numeric_transformer, numeric_cols),
    ('cat', categorical_transformer, categorical_cols)
], remainder='drop')

# Create pipeline
pipeline = Pipeline([
    ('preprocess', preprocessor),
    ('model', RandomForestClassifier(random_state=42))
])

# Use in cross-validation (no leakage!)
from sklearn.model_selection import cross_val_score
scores = cross_val_score(pipeline, X, y, cv=5)
```
