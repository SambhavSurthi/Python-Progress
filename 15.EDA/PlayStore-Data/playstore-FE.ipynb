# Cleaning Size

```python
dataset_copy.head()
```

```python
dataset_copy[dataset_copy['Size'].isna()]
```

```python
dataset_copy['Size'].unique()
```

```python
import numpy as np

def convert_size(x):
    if isinstance(x, str):
        if x.endswith('M'):
            return float(x.replace('M','')) * 1024   # MB → KB
        if x.endswith('k'):
            return float(x.replace('k',''))          # already KB
    return np.nan

dataset_copy['Size'] = dataset_copy['Size'].replace('Varies with device', np.nan)
dataset_copy['Size'] = dataset_copy['Size'].apply(convert_size)
dataset_copy['Size'] = dataset_copy['Size'].astype(float)
```
