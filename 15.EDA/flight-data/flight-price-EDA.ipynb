# Cleaning Duration

```python
dataset.head()
```

```python
dataset['Duration'].str.split(' ').str[1].str.strip('m').fillna(00)
```

```python
# dataset=(dataset.assign(
#     duration_hour=dataset['Duration'].str.split(' ').str[0].str.strip('h').fillna(00),
#     duration_min=dataset['Duration'].str.split(' ').str[1].str.strip('m').fillna(00)
# )
# .drop(columns=['Duration'])
# )

dataset = (
    dataset
    .assign(
        duration_hour = dataset['Duration']
            .str.extract(r'(\d+)h')[0]
            .fillna(0)
            .astype(int),

        duration_min = dataset['Duration']
            .str.extract(r'(\d+)m')[0]
            .fillna(0)
            .astype(int)
    )
    .drop(columns=['Duration'])
)
```

```python
encoder=OneHotEncoder()
encoded=encoder.fit_transform(dataset[['Airline','Source','Destination','Additional_Info']]).toarray()
```

```python
encoded_df=pd.DataFrame(data=encoded,columns=encoder.get_feature_names_out())
```

```python
encoded_df
```

```python
dataset=pd.concat([dataset,encoded_df],axis=1)
```

```python
dataset=dataset.drop(columns=['Airline','Source','Destination','Additional_Info'])
```
