## 1️⃣ Why do we need Regularization?

In regression, we try to learn:

$$
y = \beta_0 + \beta_1x_1 + \beta_2x_2 + \dots + \beta_nx_n
$$

If:

* too many features
* noisy data
* correlated features

👉 the model **memorizes** instead of learning patterns
This is called **overfitting**.

---

## 2️⃣ Overfitting vs Underfitting

| Case         | What happens      | R² Train  | R² Test |
| ------------ | ----------------- | --------- | ------- |
| Underfitting | Model too simple  | Low       | Low     |
| Good fit     | Balanced          | High      | High    |
| Overfitting  | Model too complex | Very High | Low     |

Your **0.98 train vs 0.97 test** → good, but regularization **makes it safer**.

---

## 3️⃣ What is Regularization (in plain English)

> **Regularization adds a penalty for large coefficients**

Instead of only minimizing error, we minimize:

$$
\text{Loss} = \text{Error} + \text{Penalty}
$$

This:

* shrinks coefficients
* reduces model complexity
* improves generalization

---

## 4️⃣ General Regularized Cost Function

For linear regression:

$$
\text{Cost} = \sum (y - \hat{y})^2 + \lambda \cdot \text{Penalty}
$$

Where:
- $\lambda$ (alpha) is the regularization strength  
- Higher $\lambda$ implies more penalty


---

## 5️⃣ Ridge Regression (L2 Regularization)

### 🔹 Penalty term

$$
\lambda \sum \beta_i^2
$$

### 🔹 Cost function

$$
\sum (y - \hat{y})^2 + \lambda \sum \beta_i^2
$$

### 🔹 What Ridge does

* Shrinks coefficients **towards zero**
* ❌ Never makes them exactly zero
* Keeps **all features**

### 🔹 Best when

* Many features
* Multicollinearity (features highly correlated)
* You believe **all features matter**

### 🔹 Visual intuition

* Coefficients become **small & smooth**
* Model becomes stable

---

## 6️⃣ Lasso Regression (L1 Regularization)

### 🔹 Penalty term

$$
\lambda \sum |\beta_i|
$$

### 🔹 Cost function

$$
\sum (y - \hat{y})^2 + \lambda \sum |\beta_i|
$$

### 🔹 What Lasso does

* Shrinks coefficients
* ✅ Can make coefficients **exactly zero**
* Performs **feature selection**

### 🔹 Best when

* Many irrelevant features
* You want a **simpler model**
* Interpretability matters

### 🔹 Important drawback

* If features are highly correlated → Lasso may pick **one randomly**

---

## 7️⃣ Elastic Net (L1 + L2 combined)

Elastic Net solves the weaknesses of Ridge & Lasso.

### 🔹 Penalty term

$$
\lambda_1 \sum |\beta_i| + \lambda_2 \sum \beta_i^2
$$

### 🔹 What it does

* Ridge → handles multicollinearity
* Lasso → feature selection
* Best of both worlds 🌍

### 🔹 Best when

* Many features
* High correlation
* Want stability + sparsity

---

## 8️⃣ Comparison Table (Very Important)

| Feature                   | Linear | Ridge | Lasso | Elastic Net |
| ------------------------- | ------ | ----- | ----- | ----------- |
| Regularization            | ❌      | ✅ L2  | ✅ L1  | ✅ L1 + L2   |
| Feature selection         | ❌      | ❌     | ✅     | ✅           |
| Handles multicollinearity | ❌      | ✅     | ⚠️    | ✅           |
| Coefficients = 0          | ❌      | ❌     | ✅     | ✅           |
| Stability                 | ❌      | ✅     | ⚠️    | ✅           |

---

## 9️⃣ Role of Alpha (λ)
- $\lambda = 0 \Rightarrow$ normal linear regression  
- Small $\lambda \Rightarrow$ slight regularization  
- Large $\lambda \Rightarrow$ strong penalty and underfitting



👉 **Always tune alpha using cross-validation**
