# Support vector machine

```python
# Import all the necessary libraries
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as plt

from sklearn.model_selection import train_test_split,GridSearchCV,StratifiedKFold
from sklearn.datasets import make_classification
from sklearn.svm import SVC

from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
```

```python
# Makke dataset
X,y=make_classification(n_samples=1000,n_features=2,n_classes=2,
                        n_clusters_per_class=1,n_redundant=0,random_state=4)
```

```python
# Convert X to Dataframe
X=pd.DataFrame(X)
```

```python
X[0]
```

```python
sns.scatterplot(x=X[0],y=X[1],hue=y)
# The follwing can be linearly seprable
```

```python
# Now we can apply SVC On the following Dataset
# Train Test Split on the dateset

X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.30,random_state=42)
```

```python
X_train
```

```python
# Train the model With Some Hyper Praramter tuning using gridsearchcv
params={'kernel':['linear', 'poly', 'rbf', 'sigmoid'],
        'C': [0.1, 1, 10, 100, 1000],
        'gamma': [1, 0.1, 0.01, 0.001, 0.0001],
        'decision_function_shape':['ovo', 'ovr']}
svc=SVC()
cv=StratifiedKFold()
grid=GridSearchCV(estimator=svc,param_grid=params,scoring='accuracy',refit=True,cv=cv)
```

```python
grid.fit(X_train,y_train)
```

```python
grid.best_estimator_
```

```python
grid.best_score_
```

```python
y_predicted=grid.predict(X_test)

# Print the metrices

print(f'accuracy_score: {accuracy_score(y_pred=y_predicted,y_true=y_test)}')
print(f'classification_report:\n {classification_report(y_pred=y_predicted,y_true=y_test)}\n')
print(f'confusion Matrix:\n {confusion_matrix(y_pred=y_predicted,y_true=y_test)}')
```
