# gaussian Naive bayes on the tips dataset

```python
# Import the necessary libraries
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px

from sklearn.preprocessing import OneHotEncoder,LabelEncoder
from sklearn.model_selection import train_test_split,GridSearchCV,StratifiedKFold,KFold
from sklearn.compose import ColumnTransformer

from sklearn.naive_bayes import GaussianNB

from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
```

```python
# Load the tips dataset


dataset=sns.load_dataset('tips')
```

```python
dataset
```

```python
dataset.columns
```

```python
dataset['time'].value_counts()
```

```python
X=dataset[['total_bill', 'tip', 'sex', 'smoker', 'time', 'size']]
```

```python
y=dataset['day']
```

```python
y
```

```python
X
```

```python
# we have 3 binary class feature and 1 multi class feature. for the binary class we implement LabelEncoding and for the multiclass we implement OHE

#Just To make Sure the Data does not leak (Dataleakage) We first perfrom train test split then implement this encoding

X_train,X_test,y_train,y_test=train_test_split(X,y,random_state=42,test_size=0.30)
```

```python
y_train
```

```python
lb1=LabelEncoder()
lb2=LabelEncoder()
lb4=LabelEncoder()

```

```python


X_train['sex']=lb1.fit_transform(X_train['sex'])
X_train['smoker']=lb2.fit_transform(X_train['smoker'])
X_train['time']=lb4.fit_transform(X_train['time'])

# Not A best preactise always use label encoder only for output features, for input use only OHE
```

```python
X_test['sex']=lb1.fit_transform(X_test['sex'])
X_test['smoker']=lb2.fit_transform(X_test['smoker'])
X_test['time']=lb4.fit_transform(X_test['time'])
```

```python
X_train
```

```python
gnb=GaussianNB()
```

```python
gnb.fit(X_train,y_train)
```

```python
y_predicted=gnb.predict(X_test)
```

```python
print(f'accuracy_score: {accuracy_score(y_pred=y_predicted,y_true=y_test)}')
print(f'classification_report:\n {classification_report(y_pred=y_predicted,y_true=y_test)}\n')
print(f'confusion Matrix:\n {confusion_matrix(y_pred=y_predicted,y_true=y_test)}')
```
