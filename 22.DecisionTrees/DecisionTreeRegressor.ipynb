# Implementating Decision Tree Regressor On the Diabetes Dataset from sklearn

```python
# Importing necessary librariess

import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split,GridSearchCV,KFold

from sklearn.tree import DecisionTreeRegressor,plot_tree

from sklearn.metrics import r2_score,mean_squared_error
```

```python
# load the dataset
dataset=load_diabetes()
```

```python
dataset
```

```python
# print the dataset describtion
print(dataset['DESCR'])
```

```python
X=pd.DataFrame(data=dataset['data'],columns=['age',
  'sex',
  'bmi',
  'bp',
  's1',
  's2',
  's3',
  's4',
  's5',
  's6'])

y=pd.Series(dataset['target'])
```

```python
X
```

```python
y
```

```python
# We dont need any Featuring Engineeing or EDA.
# Train Test split for the dataset
X_train,X_test,y_train,y_test=train_test_split(X,y,random_state=42,test_size=0.20)
```

```python
X_train
```

```python
y_train
```

```python
# Now we create model with Hyperparameters
model=DecisionTreeRegressor()
param={
    'criterion':['squared_error','friedman_mse','absolute_error'],
    'splitter':['best','random'],
    'max_depth':[1,2,3,4,5,10,15,20,25],
    'max_features':['sqrt','log2']
}
cv=KFold()
```

```python
grid=GridSearchCV(estimator=model,param_grid=param,scoring='r2',cv=cv,refit='True')
```

```python
grid.fit(X_train,y_train)
```

```python
grid.best_estimator_
```

```python
grid.best_score_
```

```python
plt.figure(figsize=(20,20))
plot_tree(grid.best_estimator_,filled=True)
```
