# Local Outlier Factory

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

# Normal data (cluster)
X_normal = np.random.randn(300, 2)

# Outliers
X_outliers = np.random.uniform(low=-8, high=8, size=(20, 2))

# Combine dataset
X = np.vstack((X_normal, X_outliers))
```

```python
plt.scatter(X[:,0],X[:,1])
```

```python
from sklearn.neighbors import LocalOutlierFactor

lof = LocalOutlierFactor(
    n_neighbors=20,
    contamination=0.06   # approx % of outliers
)

y_pred = lof.fit_predict(X)
```

```python
plt.scatter(X[y_pred == 1, 0], X[y_pred == 1, 1], c='blue', label='Normal')
plt.scatter(X[y_pred == -1, 0], X[y_pred == -1, 1], c='red', label='Outlier')
plt.legend()
plt.title("Local Outlier Factor Detection")
plt.show()
```

```python
lof_scores = lof.negative_outlier_factor_
```
