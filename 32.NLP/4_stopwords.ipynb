```python
import nltk
from nltk.tokenize import word_tokenize,sent_tokenize
from nltk.stem import WordNetLemmatizer
from nltk.corpus import stopwords
```

```python
corpus='''
Technology has transformed the way people live, work, and communicate. Over the past two decades, digital platforms have reshaped industries, enabling businesses to operate at a global scale. Remote collaboration tools allow teams to work across continents, while cloud computing provides scalable infrastructure for startups and enterprises alike.

Artificial intelligence is becoming a central component of modern systems. From recommendation engines to autonomous vehicles, machine learning models are used to analyze patterns, make predictions, and automate decision-making processes. As these systems grow more advanced, ethical considerations such as fairness, transparency, and accountability become increasingly important.

Education is also evolving rapidly. Online learning platforms provide access to knowledge for millions of learners worldwide. Personalized learning systems adapt to individual student needs, offering tailored feedback and adaptive assessments. However, access to reliable internet and digital devices remains uneven, creating disparities in educational opportunities.

Climate change presents one of the most pressing challenges of our time. Governments, corporations, and communities are investing in renewable energy solutions, sustainable agriculture, and carbon reduction strategies. Innovation in battery storage, electric vehicles, and smart grids plays a critical role in reducing greenhouse gas emissions.

Healthcare is benefiting from advances in biotechnology and data science. Wearable devices track vital signs in real time, enabling proactive health monitoring. Predictive analytics assists physicians in diagnosing diseases earlier and developing targeted treatment plans. At the same time, data privacy and regulatory compliance remain essential considerations.

Urbanization continues to shape economic development. Smart cities leverage sensors, IoT networks, and data analytics to improve transportation systems, reduce energy consumption, and enhance public safety. As populations grow, sustainable infrastructure planning becomes increasingly vital.

The future of work is being redefined by automation and digital transformation. While some roles may become obsolete, new opportunities emerge in fields such as data science, cybersecurity, and AI engineering. Continuous learning and adaptability are crucial for professionals navigating this changing landscape.


'''
```

```python
nltk.download('stopwords')
```

```python
stopwords.words('english')
```

```python
lemmatizer = WordNetLemmatizer()
stop_words = set(stopwords.words('english'))

tokens = word_tokenize(corpus)

for word in tokens:
    if word.lower() not in stop_words and word.isalpha():
        lemma = lemmatizer.lemmatize(word.lower())
        print(lemma)
```

```python
sentences=sent_tokenize(corpus)
```

```python
for i in range(len(sentences)):
    #sentences[i]=sentences[i].lower()
    words=nltk.word_tokenize(sentences[i])
    words=[lemmatizer.lemmatize(word.lower(),pos='v') for word in words if word not in set(stopwords.words('english'))]
    sentences[i]=' '.join(words)# converting all the list of words into sentences
```
