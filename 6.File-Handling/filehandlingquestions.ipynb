### Assignment 2: Writing to a File

Write a function that writes a list of strings to a file named `output.txt`, with each string on a new line.

```python
listOfStrings=['hii this is sambhav','i am a dev','i am sleepyyy']

with open('output.txt','w') as file:
    file.writelines(map(lambda x:x+'\n',listOfStrings))

with open('output.txt','r') as reader:
    content=reader.read()

print(content)
```

### Assignment 8: Counting Lines, Words, and Characters

Write a function that reads the contents of a file named `stats.txt` and returns the number of lines, words, and characters in the file.



```python
with open('stats.txt','r') as reader:
    content=reader.read()

    lines=content.split('\n')
    print(f"Lines: {len(lines)}")
    words=[i.split(' ') for i in lines]
    word=[]
    [word.append(j) for i in words for j in i]
    print(f"Words: {len(word)}")
    chars=sum([len(i) for i in word])
    print(f'Chars: {chars}')
```

### Assignment 9: Merging Multiple Files

Write a function that merges the contents of multiple files into a single file named `merged.txt`.



```python
def mergeFiles(args,outputFile):
    with open(outputFile,'w') as writer:
        for files in args:
            with open(files,'r') as reader:
                writer.write(reader.read()+'\n')


mergeFiles(['data.txt','destination.txt'],"merged.txt")
```

### Assignment 10: Splitting a Large File

Write a function that splits a large file named `large.txt` into smaller files of 100 lines each.



```python
def splitter(file,lines,outputFile):
    with open(file,'r') as reader:
        content=reader.read()
        data=list(content.split('\n'))
        print(len(data))
        with open(outputFile,'w') as writer:
            writer.writelines(map(lambda x:x+'\n', data[:lines]))

splitter('large.txt',100,'splitted.txt')
```

### Assignment 13: CSV File Operations

Write a function that reads a CSV file named `data.csv` and prints its contents as a list of dictionaries.



```python
import csv

rows=[
    ['name','marks'],
    ['sambhav',25],
    ['vaibhav',100]
]

with open('dat.csv','w',newline="") as writer:
    writer=csv.writer(writer)
    writer.writerows(rows)

with open('dat.csv','r') as reader:
    content=csv.DictReader(reader)
    print(list(content))
    
    
    
```

### Assignment 14: JSON File Operations

Write a function that reads a JSON file named `data.json` and prints its contents as a Python dictionary.


```python
import json

data={
    'name':'sambhav',
    'age':'20'
}

with open('data.json','w') as jsonWriter:
    writer=json.dump(data,jsonWriter)

with open('data.json','r') as reader:
    content=json.load(reader)
    print(content)
```
