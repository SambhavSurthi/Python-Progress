### Assignment 6: Merging Dictionaries

Create two dictionaries: one with keys as the first 5 positive integers and values as their squares, and another with keys as the next 5 positive integers and values as their squares. Merge these dictionaries into a single dictionary and print it.

```python
dict1={i:i**2 for i in range(1,6)}
print(dict1)

dict2={i:i**2 for i in range(6,11)}
print(dict2)

dict_merge={**dict1,**dict2}
print(dict_merge)
```

### Assignment 10: Dictionary and List Conversion

Create a dictionary with the first 5 positive integers as keys and their squares as values. Convert the dictionary to a list of tuples and print it.

```python
con_dict={i:i**2 for i in range(1,6)}
print(con_dict)

lst=[(k,v) for k,v in con_dict.items()]
print(lst)
```

### Assignment 11: Dictionary Filtering

Create a dictionary with the first 10 positive integers as keys and their squares as values. Create a new dictionary containing only the key-value pairs where the key is even. Print the new dictionary.

```python
flter_dict={i:i**2 for i in range(1,11)}

print(flter_dict)
new_dict={k:v for k,v in flter_dict.items() if k%2==0}
print(new_dict)
```

### Assignment 12: Dictionary Key and Value Transformation

Create a dictionary with the first 5 positive integers as keys and their squares as values. Create a new dictionary with keys and values swapped. Print the new dictionary.

```python
new_dict={i:i**2 for i in range(1,6)}
print(new_dict)

swap_dict={v:k for k,v in new_dict.items()}
print(swap_dict)
```

### Assignment 13: Default Dictionary

Create a default dictionary where each key has a default value of an empty list. Add some elements to the lists and print the dictionary.

```python
from collections import defaultdict

default_dict = defaultdict(list)
default_dict['a'].append(1)
default_dict['a'].append(2)
default_dict['b'].append(3)
print(default_dict)
```

### Assignment 14: Counting with Dictionaries

Write a function that takes a string and returns a dictionary with the count of each character in the string. Print the dictionary.

```python

str='Sambhavsurthi'
str=str.lower()
print(str)
counter_dict={}

for c in str:
    counter_dict[c]=counter_dict.get(c,0)+1

print(counter_dict)
```

```python
def remove_duplicates(lst):
    res=[]
    for i in range(0,len(lst)):
        if lst[i] not in res:
            res.append(lst[i])
    return res


lst = [1, 2, 2, 3, 4, 4, 5,1]
print(remove_duplicates(lst))
```
